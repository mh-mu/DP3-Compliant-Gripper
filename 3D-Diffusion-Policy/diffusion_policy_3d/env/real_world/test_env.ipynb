```python
from dummy_env import DummyEnv
from icecream import ic
import matplotlib.pyplot as plt
import numpy as np
```

```python

# Get the rendered image from the environment
goal_pos = (20, 60)  # Center of the larger grid
env = DummyEnv(goal_pos, 10)
image = env.render()
ic(image.shape)

# Render the initial environment
plt.imshow(image.astype('float32'))
plt.title('Environment Render')
plt.axis('off')

# Plot the force as a vector at the center of the display
center_pos = (image.shape[1] // 2, image.shape[0] // 2)
force = env.get_force()
ic(force)
plt.quiver(center_pos[0], center_pos[1], force[0], force[1], scale=10, scale_units='xy', angles='xy', color='r')
plt.show()

# Take an action and render the environment again
env.take_action((40, 200))
image = env.render()
plt.imshow(image.astype('float32'))
plt.title('Environment Render')
plt.axis('off')

# Plot the force as a vector again
force = env.get_force()
ic(force)
plt.quiver(center_pos[0], center_pos[1], force[0], force[1], scale=10, scale_units='xy', angles='xy', color='r')
plt.show()

# Print goal and robot positions
print("Goal Position:", env.get_goal_position())
```

```python

goal_pos = np.array([20, 60])  # Center of the larger grid
env = DummyEnv(goal_pos, 10)
image = env.render()
print("Max value in the image array:", np.max(image))

# Render the initial environment
plt.imshow(image.astype('float32'))
plt.title('Environment Render')
plt.axis('off')

for _ in range(10):
    action = env.calculate_action_towards_goal(initial_action_size=200)
    ic(action)
    env.take_action(action)
    image = env.render()
    
    # Render the environment after each action
    plt.imshow(image.astype('float32'))
    plt.title('Environment Render')
    plt.axis('off')
    plt.show()
```
